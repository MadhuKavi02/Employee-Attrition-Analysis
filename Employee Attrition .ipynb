```python
df=df.drop(['EmployeeCount', 'Over18', 'StandardHours'],axis=1)
```

```python
df.columns
```

Convert Target Column

```python
df['Attrition']=df['Attrition'].map({'Yes':1 ,'No':0})
```

```python
df['Attrition']
```

```python
#Real Analysis
#Attrition Count
Attrition_Count= df['Attrition'].value_counts()
print('Attrition Count: \n' ,Attrition_Count.to_string())

#Attrition Percentage
Attrition_Percentage=df['Attrition'].mean()*100
print('\nAttrition Percentage: \nPercentage of employees who left the company:' ,Attrition_Percentage)

#Department vs Attrition
DeptAttrition=df.groupby('Department')['Attrition'].mean()
print('\nDepartment vs Attrition : \nAverage count of  employees who left the company:\n\n' ,DeptAttrition.to_string())

#Salary vs Attrition
SalAttrition=df.groupby('Attrition')['MonthlyIncome'].mean()
print('\nSalary vs Attrition : \nBased on Salary Attrition rate:\n\n' ,SalAttrition.to_string())

#Age vs Attrition
AgeAttrition= df.groupby('Attrition')['Age'].mean()
print('\nAge vs Attrition : \nBased on Age Attrition rate:\n\n' ,AgeAttrition.to_string())

#OverTime vs Attrition
OTAttrition=df.groupby('OverTime')['Attrition'].mean()*100
print('\nOverTime vs Attrition : \nBased on OverTime Attrition rate:\n\n' ,OTAttrition.to_string())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Attrition Count Plot
plt.figure(figsize=(6,4))

ax=sns.countplot(x='Attrition',data=df,hue='Attrition',palette='Set2')

plt.title("Employee Attrition Distribution",fontsize=14,fontweight='bold')
plt.xlabel("Attrition (0 = No, 1 = Yes)")
plt.ylabel("Number of employees")

for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as mtick

plt.figure(figsize=(7,5))

ax = sns.barplot(
    x='Department',
    y='Attrition',
    data=df,
    hue='Department',      # assign hue
    palette='Set2',
    errorbar=None,         # updated parameter
    legend=False           # remove duplicate legend
)

# Title & labels
plt.title("Attrition Rate by Department", fontsize=14, fontweight='bold')
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

# Convert Y axis to %
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))

# Add labels
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height*100:.1f}%',
                (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom', fontsize=10)

plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(6,4))

ax = sns.boxplot(
    x='Attrition',
    y='MonthlyIncome',
    data=df,
    hue='Attrition',
    palette='Set2',
    legend=False
)

# Add median values
medians = df.groupby('Attrition')['MonthlyIncome'].median()

for i, median in enumerate(medians):
    ax.text(i, median, f'{int(median)}', 
            ha='center', va='bottom', fontweight='bold')

plt.title("Salary Distribution by Attrition", fontweight='bold')
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Monthly Income")

plt.show()
```

```python
#Overtime vs Attrition
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))
ax=sns.barplot(x='OverTime',y='Attrition',data=df,hue='OverTime',palette='Set2',errorbar=None,legend=False)
plt.title("Attrition rate by Overtime",fontweight='bold')
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
for p in ax.patches:
    ax.annotate(f'{p.get_height()*100:.1f}%',(p.get_x() + p.get_width()/2.,p.get_height()),ha='center',va='bottom')
plt.show
           
```

```python
#Age Group analysis visulaization
df['Age_Group']=pd.cut(df['Age'],bins=[18,25,35,60],labels=['18-25','25-35','Above 35'])
plt.figure(figsize=(6,4))
ax=sns.barplot(x='Age_Group',y='Attrition',data=df,hue='Age_Group',errorbar=None,legend=False,palette='Set2')
plt.title('Attrition rate by Age Group',fontweight='bold')
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
for p in ax.patches:
    ax.annotate(f'{p.get_height()*100:.1f}%',
    (p.get_x()+p.get_width()/2.,p.get_height()),
    ha='center',va='bottom')
plt.show()
```
